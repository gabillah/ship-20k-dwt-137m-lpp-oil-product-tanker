# Calculation Notebook

This notebook performs engineering calculations and **automatically exports the results**
into `.tex` snippet files that `report.tex` can `\input{}` directly.

## Workflow
1. Run all cells in this notebook (Kernel → Restart & Run All)
2. Snippet files (`calc_values.tex`) are written to the same folder
3. Compile `report.tex` with XeLaTeX → BIBTeX → XeLaTeX → XeLaTeX

```python
import os
import math

# ─────────────────────────────────────────────
# Helper: format a number for LaTeX
# ─────────────────────────────────────────────
def fmt(value, decimals=4):
    """Round and format a float for LaTeX output."""
    return f"{value:.{decimals}f}"

def fmt_sci(value, decimals=3):
    """Scientific notation for LaTeX, e.g. 1.234 \\times 10^{5}"""
    if value == 0:
        return "0"
    exp = int(math.floor(math.log10(abs(value))))
    mantissa = value / (10 ** exp)
    return rf"{mantissa:.{decimals}f} \times 10^{{{exp}}}"

print("Helper functions loaded.")
```

## 1. Simple Addition Example

Demonstrates the basic pattern: compute `c = a + b`, then export to LaTeX.

```python
# ── Inputs ──────────────────────────────────
a = 35   # first operand
b = 1.5   # second operand

# ── Calculation ─────────────────────────────
c = a + b

print(f"a = {a}")
print(f"b = {b}")
print(f"c = a + b = {c}")
```

## 2. Pipe Flow Calculation (Darcy-Weisbach)

Head loss in a pipe:
$$h_f = f \cdot \frac{L}{D} \cdot \frac{V^2}{2g}$$

```python
# ── Inputs ──────────────────────────────────
f_friction = 0.015   # Darcy friction factor  [-]
L_pipe     = 100.0    # pipe length            [m]
D_pipe     = 0.1    # pipe diameter          [m]
V_flow     = 2.0     # flow velocity          [m/s]
g_gravity  = 9.81    # gravitational accel.   [m/s²]

# ── Calculation ─────────────────────────────
h_f = f_friction * (L_pipe / D_pipe) * (V_flow**2 / (2 * g_gravity))

Re = (V_flow * D_pipe) / 1e-6   # Reynolds number (kinematic visc. of water ≈ 1e-6 m²/s)

print(f"Friction factor  f  = {f_friction}")
print(f"Pipe length      L  = {L_pipe} m")
print(f"Pipe diameter    D  = {D_pipe} m")
print(f"Flow velocity    V  = {V_flow} m/s")
print(f"Reynolds number  Re = {Re:.0f}")
print(f"Head loss        hf = {h_f:.4f} m")
```

## 3. Export All Values → `calc_values.tex`

Each value is written as a `\newcommand` macro.  
`report.tex` uses `\input{calc_values}` to load them, then references
them by name (e.g. `\ValC`, `\ValHf`) — no hard-coding numbers in LaTeX.

```python
# ─────────────────────────────────────────────────────────────────
# Build the snippet: one \newcommand per calculated value
# ─────────────────────────────────────────────────────────────────

# Determine output path: same directory as this notebook
notebook_dir = os.path.dirname(os.path.abspath("calculation.ipynb"))
output_path  = os.path.join(notebook_dir, "calc_values.tex")

lines = [
    r"% ============================================================",
    r"% AUTO-GENERATED by calculation.ipynb — DO NOT EDIT MANUALLY",
    r"% Re-run the notebook to regenerate this file.",
    r"% ============================================================",
    "",
    r"% ── Section 1: Simple Addition ──────────────────────────────",
    rf"\newcommand{{\ValA}}{{{fmt(a, 2)}}}",
    rf"\newcommand{{\ValB}}{{{fmt(b, 2)}}}",
    rf"\newcommand{{\ValC}}{{{fmt(c, 2)}}}",
    "",
    r"% ── Section 2: Darcy-Weisbach ───────────────────────────────",
    rf"\newcommand{{\ValFriction}}{{{fmt(f_friction, 3)}}}",
    rf"\newcommand{{\ValLpipe}}{{{fmt(L_pipe, 1)}}}",
    rf"\newcommand{{\ValDpipe}}{{{fmt(D_pipe, 2)}}}",
    rf"\newcommand{{\ValVflow}}{{{fmt(V_flow, 1)}}}",
    rf"\newcommand{{\ValRe}}{{{Re:.0f}}}",
    rf"\newcommand{{\ValHf}}{{{fmt(h_f, 4)}}}",
    "",
    r"% ── Scientific notation example ─────────────────────────────",
    rf"\newcommand{{\ValReSci}}{{{fmt_sci(Re)}}}",
]

snippet = "\n".join(lines)

with open(output_path, "w", encoding="utf-8") as f:
    f.write(snippet)

print(f"Written: {output_path}")
print("\n" + "─" * 60)
print(snippet)
```

## 4. Verify the exported file

Quick sanity check — read back and display the file contents.

```python
with open(output_path, "r", encoding="utf-8") as f:
    content = f.read()

print("Contents of calc_values.tex:")
print("=" * 60)
print(content)
print("=" * 60)
print("\n✓ File verified. Now compile report.tex with XeLaTeX.")
```
